# Visualize job openings

Same charts as the 2018 `Data Visualisations.ipynb` notebook:

- jobs by technology
- jobs by sector
- education / degree field
- years of experience
- location word cloud

Default input is the recovered 2018 Texas table. Point `DATA` at `data/jobs.csv` after a live collect.


```python
from pathlib import Path
import sys

import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path.cwd() if (Path.cwd() / "job_scraping").exists() else Path.cwd().parent
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from job_scraping.skills import analyze_jobs, category_totals
from job_scraping.visualize import (
    plot_jobs_by_education,
    plot_jobs_by_experience,
    plot_jobs_by_sector,
    plot_jobs_by_technology,
    plot_location_wordcloud,
    save_figures,
)

DATA = ROOT / "data" / "texas_jobs_2018.csv"
visual_df = analyze_jobs(pd.read_csv(DATA))
visual_df.columns[:12].tolist(), len(visual_df)
```

```python
category_totals(visual_df)
```

```python
fig, ax = plt.subplots(figsize=(8, 6))
plot_jobs_by_technology(visual_df, ax)
fig.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(8, 5))
plot_jobs_by_sector(visual_df, ax)
fig.tight_layout()
plt.show()
```

```python
fig, axs = plt.subplots(1, 2, figsize=(12, 4))
plot_jobs_by_education(visual_df, axs)
fig.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10, 6))
plot_location_wordcloud(visual_df, ax)
fig.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(7, 6))
plot_jobs_by_experience(visual_df, ax)
fig.tight_layout()
plt.show()
```

```python
paths = save_figures(visual_df, ROOT / "figures")
for path in paths:
    print(path)
```
